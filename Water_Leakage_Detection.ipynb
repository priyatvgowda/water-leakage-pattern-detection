```python
import pandas as pd
import matplotlib.pyplot as plt

# Water usage dataset
data = pd.DataFrame({
    "Hour": range(1, 11),
    "Water_Usage": [10, 12, 11, 10, 45, 50, 48, 12, 11, 10]
})

# Calculate average usage
average = data["Water_Usage"].mean()

# Set leakage threshold
threshold = average * 1.5

# Detect abnormal water usage
data["Status"] = data["Water_Usage"].apply(
    lambda x: "Possible Leakage" if x > threshold else "Normal"
)

# Display results
print("WATER LEAKAGE PATTERN DETECTION")
print(data)

# Leakage alert
if (data["Status"] == "Possible Leakage").any():
    print("\nALERT: Possible water leakage detected!")
else:
    print("\nNo abnormal water usage detected.")

# Graph
plt.figure(figsize=(8, 4))
plt.plot(data["Hour"], data["Water_Usage"], marker="o")
plt.axhline(threshold, linestyle="--", label="Leakage Threshold")
plt.xlabel("Hour")
plt.ylabel("Water Usage")
plt.title("Water Leakage Pattern Detection")
plt.legend()
plt.grid()
plt.show()
```
